```python
import json
import matplotlib.pyplot as plt


# =========================================================
# LOAD DATA
# =========================================================
with open(r"D:\NHAT MINH\CODE\BAI LAB\Vòng 5\elo_results.json") as f:
    data = json.load(f)

elo = data["elo_summary"]
correctness = data["correctness"]


# =========================================================
# 1. BAR CHART + ERROR BAR
# =========================================================
def plot_elo():
    models = list(elo.keys())
    means = [elo[m]["mean"] for m in models]
    stds = [elo[m]["std"] for m in models]

    plt.figure()
    plt.errorbar(models, means, yerr=stds, fmt='o', capsize=5)

    plt.xlabel("Models")
    plt.ylabel("Elo Score")
    plt.title("Elo with Confidence Interval")

    plt.show()


# =========================================================
# 2. DUAL PLOT (paper style)
# =========================================================
def plot_dual():
    models = list(elo.keys())

    x = [elo[m]["mean"] for m in models]
    y = [correctness[m] for m in models]

    plt.figure()

    plt.scatter(x, y)

    for i, m in enumerate(models):
        plt.text(x[i], y[i], m)

    plt.xlabel("Insightfulness (Elo)")
    plt.ylabel("Correctness")
    plt.title("Insightfulness vs Correctness")

    plt.show()


# =========================================================
# RUN
# =========================================================
plot_elo()
plot_dual()
```
